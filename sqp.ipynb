```
import numpy as np
import scipy as cp
import matplotlib.pyplot as plt
from scipy.optimize import minimize
```

```
fun = lambda x: x[0] + x[1]
```

```
cons = ({'type': 'ineq', 'fun': lambda x:  x[0]**2 + x[1]**2 - 1})
```

```
res = minimize(fun, (1, 1), method='SLSQP', constraints=cons)
```

```
res
```

```
fun(res['x'])
```

```
A = np.array([
    [2,0,-1],
    [0,2,-1],
    [1,1,0]
])
b = np.array([8, 12, 8])

np.linalg.solve(A,b)
```
